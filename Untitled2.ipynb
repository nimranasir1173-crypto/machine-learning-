```python
import zipfile
import os

zip_path = '/content/archive.zip'
extract_path = '/content/extracted_data'

with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_path)

print("Extracted files:")
for root, dirs, files in os.walk(extract_path):
    for file in files:
        print(os.path.join(root, file))
```

```python
import pandas as pd

# Load the dataset
df = pd.read_csv('/content/extracted_data/starbucks.csv')

# Print basic information
print("Dataset Shape:", df.shape)
print("\n--- Dataset Info ---")
df.info()
print("\n--- First 5 Rows ---")
df.head()
```

```python
import numpy as np

# 1. Clean column names by stripping extra spaces
df.columns = df.columns.str.strip()
print("Cleaned Column Names:", list(df.columns))

# 2. Clean and convert numeric columns
# Remove '%' and handle other string representations
cols_to_clean = [
    'Total Fat (g)',
    'Vitamin A (% DV)',
    'Vitamin C (% DV)',
    'Calcium (% DV)',
    'Iron (% DV)',
    'Caffeine (mg)'
]

for col in cols_to_clean:
    # Strip whitespace, replace percentage signs, and coerce non-numeric (like 'Varies', 'varies') to NaN
    df[col] = df[col].astype(str).str.replace('%', '', regex=False).str.strip()
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Check the updated data types and missing values summary
print("\n--- Cleaned Data Types ---")
print(df[cols_to_clean].dtypes)
print("\nMissing values count per cleaned column:")
print(df[cols_to_clean].isnull().sum())

# Get a descriptive summary of numeric fields
print("\n--- Descriptive Statistics ---")
df.describe()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Count the number of items in each beverage category
category_counts = df['Beverage_category'].value_counts()

# Plot the pie chart
plt.figure(figsize=(10, 8))
plt.pie(
    category_counts,
    labels=category_counts.index,
    autopct='%1.1f%%',
    startangle=140,
    colors=sns.color_palette('pastel', len(category_counts))
)
plt.title('Distribution of Starbucks Beverage Categories', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
from sklearn.model_selection import train_test_split

# Split the DataFrame into 80% training and 20% testing
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42)

# Display the shapes
print(f"Original dataset shape: {df.shape}")
print(f"Training dataset shape: {train_df.shape} (80%)")
print(f"Testing dataset shape: {test_df.shape} (20%)")
```

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.impute import SimpleImputer
import numpy as np

# Define target and features
target_col = 'Calories'
feature_cols = [
    'Total Fat (g)',
    'Trans Fat (g)',
    'Saturated Fat (g)',
    'Sodium (mg)',
    'Total Carbohydrates (g)',
    'Cholesterol (mg)',
    'Dietary Fibre (g)',
    'Sugars (g)',
    'Protein (g)',
    'Vitamin A (% DV)',
    'Vitamin C (% DV)',
    'Calcium (% DV)',
    'Iron (% DV)',
    'Caffeine (mg)'
]

# Extract features and targets from training and testing splits
X_train = train_df[feature_cols]
y_train = train_df[target_col]
X_test = test_df[feature_cols]
y_test = test_df[target_col]

# Handle missing values (e.g., in Caffeine or Total Fat) by imputing the median
imputer = SimpleImputer(strategy='median')
X_train_imputed = imputer.fit_transform(X_train)
X_test_imputed = imputer.transform(X_test)

# Initialize and train the Decision Tree Regressor
dt_model = DecisionTreeRegressor(max_depth=5, random_state=42)
dt_model.fit(X_train_imputed, y_train)

# Predict on training and testing sets
y_train_pred = dt_model.predict(X_train_imputed)
y_test_pred = dt_model.predict(X_test_imputed)

# Evaluate the model
train_rmse = np.sqrt(mean_squared_error(y_train, y_train_pred))
test_rmse = np.sqrt(mean_squared_error(y_test, y_test_pred))
train_r2 = r2_score(y_train, y_train_pred)
test_r2 = r2_score(y_test, y_test_pred)

print(f"--- Model Evaluation: Decision Tree Regressor ---")
print(f"Training RMSE: {train_rmse:.2f} calories")
print(f"Testing RMSE: {test_rmse:.2f} calories")
print(f"Training R² Score: {train_r2:.4f}")
print(f"Testing R² Score: {test_r2:.4f}")
```

```python
import matplotlib.pyplot as plt

# Feature Importance
importances = dt_model.feature_importances_
indices = np.argsort(importances)[::-1]

plt.figure(figsize=(10, 6))
plt.title("Feature Importances in Predicting Calories")
plt.bar(range(len(feature_cols)), importances[indices], align="center")
plt.xticks(range(len(feature_cols)), [feature_cols[i] for i in indices], rotation=90)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 6))
sns.scatterplot(x=y_test, y=y_test_pred, alpha=0.8, color='teal', edgecolor='black', s=80)

# Line of perfect prediction
max_val = max(y_test.max(), y_test_pred.max())
min_val = min(y_test.min(), y_test_pred.min())
plt.plot([min_val, max_val], [min_val, max_val], color='red', linestyle='--', linewidth=2, label='Perfect Prediction')

plt.title('Decision Tree Model: Actual vs. Predicted Calories', fontsize=14, fontweight='bold')
plt.xlabel('Actual Calories', fontsize=12)
plt.ylabel('Predicted Calories', fontsize=12)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
import joblib

# Save the model to a file
model_filename = 'decision_tree_model.joblib'
joblib.dump(dt_model, model_filename)

# Save the fitted imputer as well to handle missing values on new inputs
imputer_filename = 'simple_imputer.joblib'
joblib.dump(imputer, imputer_filename)

print(f"Model successfully saved to: {model_filename}")
print(f"Imputer successfully saved to: {imputer_filename}")
```
